```python
import pandas as pd
import torch
import matplotlib.pyplot as plt
import  seaborn as sns
import numpy as np
from torch import nn
```

```python
# Data set
from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()
x=data.data
y=data.target
df=pd.DataFrame(x,columns=data.feature_names)
df['target']=y
df['target'].value_counts()
```

```python
# Normalize the features
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Re-split data to ensure X_train and X_test are correctly initialized as ndarrays
X_train,X_test, Y_train, Y_test= train_test_split(x,y,train_size=0.95,random_state=42)

scalar = StandardScaler()
X_train=scalar.fit_transform(X_train)
X_test=scalar.transform(X_test)
```

```python
# converting to the tensors
X_train=torch.tensor(X_train,dtype=torch.float32)
X_test=torch.tensor(X_test,dtype=torch.float32)
Y_train=torch.tensor(Y_train,dtype=torch.float32)
Y_test=torch.tensor(Y_test,dtype=torch.float32)
```

```python
# setting up gpu
device='cuda' if torch.cuda.is_available() else 'cpu'
device
```

```python
# changing tensor to gpu
X_train=X_train.to(device)
X_test=X_test.to(device)
Y_train=Y_train.to(device)
Y_test=Y_test.to(device)
```

```python
# creating the simple model
model=nn.Sequential(
    nn.Linear(in_features=X_train.shape[1],out_features=10),
    nn.ReLU(),
    nn.Linear(in_features=10,out_features=5),
    nn.ReLU(),
    nn.Linear(in_features=5,out_features=2)


)
```
